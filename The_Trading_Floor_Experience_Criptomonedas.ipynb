```python
import yfinance as yf

btc = yf.download('BTC-USD',
  start='2024-01-01',
  end='2024-12-31',
  interval='1wk')

eth = yf.download('ETH-USD',
  start='2024-01-01',
  end='2024-12-31',
  interval='1wk')

sol = yf.download('SOL-USD',
  start='2024-01-01',
  end='2024-12-31',
  interval='1wk')

ltc = yf.download('LTC-USD',
  start='2024-01-01',
  end='2024-12-31',
  interval='1wk')
```
