**RENDIMIENTO**

Para comprender el atractivo de cada activo, hemos calculado los **rendimientos semanales** de cada criptomoneda. Este indicador nos permite cuantificar la ganancia o pérdida porcentual de una inversión en un período determinado, siendo fundamental para comparar la performance individual de cada activo.

La fórmula empleada es la del cambio porcentual sobre el precio de cierre, lo que nos da una medida precisa de la fluctuación del valor a lo largo del tiempo.

```python
# btc['Close'] tiene los precios de cierre, ahora accedemos correctamente con MultiIndex
btc['retorno'] = btc[('Close', 'BTC-USD')].pct_change()
eth['retorno'] = eth[('Close', 'ETH-USD')].pct_change()
sol['retorno'] = sol[('Close', 'SOL-USD')].pct_change()
ada['retorno'] = ada[('Close', 'ADA-USD')].pct_change()
```

### Evolución de Rendimientos Acumulados: Gráfico de Líneas

Este gráfico visualiza la trayectoria del rendimiento de una inversión inicial (base 1) para cada una de las criptomonedas (BTC, ETH, SOL, ADA) a lo largo del tiempo. Es una herramienta clave para entender cómo una inversión en cada activo ha crecido (o decrecido) a lo largo del período analizado.

```python
import plotly.express as px

# Calcular los rendimientos acumulados si no se ha hecho aún en esta sesión
# Esto es un respaldo, ya deberían estar calculados en 'rendimientos_acumulados_df'
if 'rendimiento_acumulado' not in btc.columns:
    btc['rendimiento_acumulado'] = (1 + btc['retorno']).cumprod()
    eth['rendimiento_acumulado'] = (1 + eth['retorno']).cumprod()
    sol['rendimiento_acumulado'] = (1 + sol['retorno']).cumprod()
    ada['rendimiento_acumulado'] = (1 + ada['retorno']).cumprod()

    rendimientos_acumulados_df = pd.DataFrame({
        'BTC': btc['rendimiento_acumulado'],
        'ETH': eth['rendimiento_acumulado'],
        'SOL': sol['rendimiento_acumulado'],
        'ADA': ada['rendimiento_acumulado']
    })

# Graficar los rendimientos acumulados
fig_rendimientos_acumulados = px.line(rendimientos_acumulados_df, x=rendimientos_acumulados_df.index, y=rendimientos_acumulados_df.columns,
                                           title='Retorno Acumulado de Criptomonedas (2025-2026)',
                                           labels={'value':'Retorno Acumulado (Base 1)', 'Date':'Fecha', 'variable':'Criptomoneda'},
                                           hover_data={'value':':,.2f'})

fig_rendimientos_acumulados.update_layout(
    xaxis_title='Fecha',
    yaxis_title='Retorno Acumulado (Base 1)',
    hovermode='x unified'
)

fig_rendimientos_acumulados.show()
```

**Eje Y** (Retorno Acumulado - Base 1): Los valores en este eje indican el factor de crecimiento. Si el valor es 1.50, significa que tu inversión inicial de 1 unidad ahora vale 1.50 (un retorno del 50%). Si es 0.80, significa que tu inversión de 1 unidad ahora vale 0.80 (una pérdida del 20%).

**Líneas Individuales:** Cada línea de color representa una criptomoneda diferente (BTC, ETH, SOL, ADA). Al seguir una línea, puedes observar cómo su valor de inversión acumulado subió o bajó en cualquier día del período.

**Interpretación de Ganancia/Pérdida:** Al final del período, el valor en el eje Y para cada línea te da la ganancia o pérdida total de tu inversión hipotética. Por ejemplo, si la línea de BTC termina en 0.86, significa que tu inversión inicial de 1 ahora vale 0.86, lo que implica una pérdida del 14%.

**RIESGO: VOLATILIDAD ANUALIZADA**

En un mercado tan dinámico como el de las criptomonedas, la gestión del riesgo es tan crítica como la búsqueda de rendimientos. Hemos cuantificado este riesgo mediante la **volatilidad anualizada**, una métrica estándar que mide la dispersión de los precios de un activo alrededor de su media. En términos más simples, nos indica cuánto 'se mueve' el precio de una criptomoneda en un año.

Una mayor volatilidad implica un mayor riesgo potencial, pero también la oportunidad de mayores retornos. Nuestro objetivo es equilibrar estos factores inteligentemente. El cálculo se realiza a partir de la desviación estándar de los rendimientos semanales, proyectada a un horizonte anual.

```python
import numpy as np

# Ajustado a 252 días hábiles para volatilidad anualizada
vol_anual_btc = btc['retorno'].std() * np.sqrt(252)
vol_anual_eth = eth['retorno'].std() * np.sqrt(252)
vol_anual_sol = sol['retorno'].std() * np.sqrt(252)
vol_anual_ada = ada['retorno'].std() * np.sqrt(252)
```

```python
import pandas as pd

# Crear un DataFrame con las volatilidades anualizadas
volatilidades = pd.DataFrame({
    'Criptomoneda': ['BTC', 'ETH', 'SOL', 'ADA'],
    'Volatilidad Anualizada': [vol_anual_btc, vol_anual_eth, vol_anual_sol, vol_anual_ada]
})

# Mostrar el DataFrame de volatilidades
display(volatilidades)

# Crear el gráfico de barras con Plotly Express
fig_volatilidad = px.bar(volatilidades, x='Criptomoneda', y='Volatilidad Anualizada',
                         title='Volatilidad Anualizada de Criptomonedas (2025-2026)',
                         color='Criptomoneda',  # Colorea las barras por criptomoneda
                         labels={'Volatilidad Anualizada': 'Volatilidad Anualizada'},
                         text='Volatilidad Anualizada') # Muestra el valor en la barra
fig_volatilidad.update_traces(texttemplate='%{text:.2f}', textposition='outside') # Formatear el texto
fig_volatilidad.update_layout(uniformtext_minsize=8, uniformtext_mode='hide') # Ajuste de texto
fig_volatilidad.show()
```

### Máxima Caída (Maximum Drawdown - MDD)

```python
import numpy as np
import pandas as pd

def calculate_mdd(cumulative_returns):
    if cumulative_returns.empty or cumulative_returns.isnull().all():
        return np.nan
    # Asegurarse de que no haya NaN en el cálculo del MDD si se pueden omitir los primeros
    temp_returns = cumulative_returns.dropna()
    if temp_returns.empty:
        return np.nan

    # Calcula el máximo histórico (peak) hasta la fecha actual
    running_max = temp_returns.cummax()
    # Calcula la caída actual desde el máximo histórico
    drawdown = (temp_returns / running_max) - 1
    # El Maximum Drawdown es la caída más grande
    mdd = drawdown.min()
    return mdd

# Calcular el MDD para cada criptomoneda
mdd_btc = calculate_mdd(btc['rendimiento_acumulado'])
mdd_eth = calculate_mdd(eth['rendimiento_acumulado'])
mdd_sol = calculate_mdd(sol['rendimiento_acumulado'])
mdd_ada = calculate_mdd(ada['rendimiento_acumulado'])

# Crear un DataFrame para mostrar los resultados
mdd_df = pd.DataFrame({
    'Criptomoneda': ['BTC', 'ETH', 'SOL', 'ADA'],
    'Maximum Drawdown': [mdd_btc, mdd_eth, mdd_sol, mdd_ada]
})

# Mostrar el DataFrame
display(mdd_df)

# Crear un gráfico de barras para el MDD
fig_mdd = px.bar(mdd_df, x='Criptomoneda', y='Maximum Drawdown',
                 title='Maximum Drawdown (Peor Caída Histórica) por Criptomoneda',
                 labels={'Maximum Drawdown':'MDD', 'Criptomoneda':'Criptomoneda'},
                 color='Criptomoneda', # Colorea las barras por criptomoneda
                 text='Maximum Drawdown') # Muestra el valor en la barra
fig_mdd.update_traces(texttemplate='%{text:.2%}', textposition='outside') # Formatear el texto como porcentaje
fig_mdd.update_layout(uniformtext_minsize=8, uniformtext_mode='hide') # Ajuste de texto
fig_mdd.show()
```

Este gráfico de barras muestra la *Máxima Caída (Maximum Drawdown - MDD)* para cada criptomoneda. El MDD es una métrica clave de riesgo que representa la mayor pérdida porcentual desde un pico hasta un valle subsiguiente antes de alcanzar un nuevo pico. En otras palabras, mide la peor pérdida que un inversor habría sufrido si hubiera comprado en el pico y vendido en el valle más bajo después de ese pico.

**Interpretación:**

*   Un MDD de -0.50 (-50%) significa que, en algún momento, la criptomoneda perdió la mitad de su valor desde su máximo histórico hasta ese punto.
*   Cuanto mayor (menos negativo) sea el MDD, menor fue la peor caída, lo que indica una mayor resiliencia del activo.

Esta visualización es fundamental para entender el riesgo histórico de una inversión, mostrando el peor escenario de pérdida que se ha presentado en el periodo analizado.

### Porcentaje de Días con Rendimiento Negativo (Frecuencia de Pérdidas)

Esta métrica nos muestra la proporción de días en los que el rendimiento de una criptomoneda fue negativo, lo que nos da una idea directa de la frecuencia de las pérdidas. Es un indicador clave para evaluar el riesgo asociado a la volatilidad a la baja.

```python
import pandas as pd
import plotly.express as px

# Calcular el porcentaje de días con rendimiento negativo para cada criptomoneda
neg_ret_btc = (btc['retorno'] < 0).sum() / btc['retorno'].dropna().count()
neg_ret_eth = (eth['retorno'] < 0).sum() / eth['retorno'].dropna().count()
neg_ret_sol = (sol['retorno'] < 0).sum() / sol['retorno'].dropna().count()
neg_ret_ada = (ada['retorno'] < 0).sum() / ada['retorno'].dropna().count()

# Crear un DataFrame para mostrar los resultados
neg_returns_df = pd.DataFrame({
    'Criptomoneda': ['BTC', 'ETH', 'SOL', 'ADA'],
    'Porcentaje Días Negativos': [neg_ret_btc, neg_ret_eth, neg_ret_sol, neg_ret_ada]
})

# Mostrar el DataFrame
display(neg_returns_df)

# Crear un gráfico de barras para el porcentaje de días negativos
fig_neg_returns = px.bar(neg_returns_df, x='Criptomoneda', y='Porcentaje Días Negativos',
                         title='Porcentaje de Días con Rendimiento Negativo (2025-2026)',
                         labels={'Porcentaje Días Negativos':'% Días Negativos', 'Criptomoneda':'Criptomoneda'},
                         color='Criptomoneda', # Colorea las barras por criptomoneda
                         text='Porcentaje Días Negativos') # Muestra el valor en la barra
fig_neg_returns.update_traces(texttemplate='%{text:.2%}', textposition='outside') # Formatear el texto como porcentaje
fig_neg_returns.update_layout(uniformtext_minsize=8, uniformtext_mode='hide') # Ajuste de texto
fig_neg_returns.show()
```

Este gráfico de barras muestra claramente la **frecuencia con la que cada criptomoneda ha tenido un rendimiento diario negativo**. Una barra más alta indica que el activo ha registrado pérdidas con mayor asiduidad durante el periodo analizado.

**Interpretación para inversionistas:**

*   Un alto porcentaje de días negativos sugiere una mayor inestabilidad y un riesgo más frecuente de ver el valor de la inversión disminuir en el corto plazo. Esto es importante para aquellos inversionistas con menor tolerancia al riesgo o que planean realizar operaciones frecuentes.
*   Un bajo porcentaje de días negativos, por el contrario, indica una mayor consistencia en el rendimiento positivo, lo que puede ser percibido como una señal de mayor estabilidad, aunque no necesariamente de mayor rentabilidad absoluta.

Esta métrica complementa la volatilidad y el Maximum Drawdown, ofreciendo una perspectiva sobre la **constancia del riesgo de pérdida**, en lugar de solo la magnitud de las caídas.
