## Paso 2 – Qué es un *ticker*

>  “Cuando en finanzas hablamos de un *ticker*, nos referimos al código con el que se identifica una acción en la bolsa.”

| Empresa | Ticker |
|----------|---------|
| Apple Inc. | AAPL |
| Microsoft | MSFT |
| Google (Alphabet) | GOOG |
| Tesla | TSLA |

>  *Reflexión:* así como una cédula identifica a una persona, el *ticker* identifica una acción.


**bold text**
## Paso 3 – Definir nuestro portafolio


```python

tickers = ["AAPL", "NFLX", "GOOG"]
```

```python
type(tickers)
```


>  “Cada uno de estos *tickers* representa una empresa de nuestro portafolio.”  




## Paso 4 – Descargar los datos

> “Vamos a descargar precios históricos de los últimos seis meses.”


```python

data = yf.download(tickers, start="2026-01-01", end="2026-04-15") # descargamos los datos
data.head()
```


📘 **Explicación:**
- `start` y `end` definen el rango de tiempo.
- El resultado es una tabla con las columnas `Open`, `High`, `Low`, `Close` y `Volume`.





## Paso 5 – Trabajar solo con precios de cierre


```python

close_prices = data["Close"]
close_prices.head()
```


>  “El precio de cierre es el más importante, porque refleja el valor final de cada jornada bursátil.”  
 Así es como vemos la evolución diaria de las tres empresas.



##  Paso 6 – Visualizar la evolución de precios


```python

close_prices.plot(figsize=(10,5))
plt.title("Evolución de precios de cierre - Portafolio tecnológico")
plt.xlabel("Fecha")
plt.ylabel("Precio (USD)")
plt.grid(True)
plt.show()
```


>  *Explicación:*  
> - Cada línea representa una empresa.  
> - Nos muestra cómo ha cambiado su valor en el tiempo.  

 **Preguntas:**  
- “¿Cuál de las empresas parece tener el precio más alto?”  
- “¿Observas alguna tendencia clara al alza o baja?”



## 📉 Paso 7 – Analizar la distribución de precios


```python

close_prices.hist(figsize=(10,5), bins=20)
plt.suptitle("Distribución de precios de cierre - Portafolio")
plt.show()
```


> “Este gráfico nos muestra cuántas veces los precios se concentraron en ciertos valores.”

 *Interpretación:*  
- Una distribución muy extendida indica alta **volatilidad**.  
- Una distribución más compacta indica **estabilidad**.
